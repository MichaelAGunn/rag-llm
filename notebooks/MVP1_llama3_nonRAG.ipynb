# Biblical RAG LLM MVP 1

This notebook tests llama3's ability to with no Bible support.

## Instantiate LLM

```python
from langchain_ollama import OllamaLLM

llm = OllamaLLM(model='llama3')
print("Model isntantiated.")
```

## Test the RAG Model

```python
questions = [
    "Where did Jesus talk about wine?",
    "Who was Abraham's son?",
    "Where did God send Jonah? And where did Jonah try to go instead?",
    "Who are Luke's books addressed to?",
    "Which psalm says that angels will protect you even from dashing your foot against a stone?",
    "Why did God regret making Saul the King of Israel?",
    "What did Paul disagree with Peter about according to Paul's letter to the Galatians? Who was right?",
    "Why did Peter feel comfortable eating pork and shellfish after receiving the Holy Spirit?",
    "When Jesus mentions that Moses lifted the serpent in the wilderness (John 3:14), what is He referencing?",
    "What is the genealogy of Enoch going back to Adam?",
    "Why did God flood the world in the time of Noah?",
    "What does Messiah mean in Hebrew?",
    "How do you spell 'Jesus' in Hebrew and Greek?",
    "What are the similarities between Daniel's prophesies and Revelation's prophesies?"
]
print("There are {} questions in the test suite.".format(len(questions)))

for question in questions:
    answer = llm.invoke(question)
    qa = {'question': question, 'answer': answer}
    print(qa)
```

## Conclusion

### Where did Jesus talk about wine?
The model mentioned the wine miracle at Cana, the mention of wine in the last supper, and the comparison between his teaching and wine. The list is incomplete.
### Who was Abraham's son?
The model mentions both of Abraham's sons and their mothers.
### Where did God send Jonah? And where did Jonah try to go instead?
The model correctly answers both questions (the answers are Nineveh and Tarshish). And then it goes on to tell half of the story in the Book of Jonah.
### Who are Luke's books addressed to?
The model currectly mentions Theophilus, and goes into great detail about the theories of the identity of Theophilus.
### Which psalm says that angels will protect you even from dashing your foot against a stone?
The model currectly identifies Psalm 91, and even knows the correct verse (12).
### Why did God regret making Saul the King of Israel?
The model currectly identifies 1 Samuel 15:35 as the exact place where it says that God regretted making Saul the King of Israel, and provides several examples of Saul's disobedience. There seems to be hallucination in the first sentence, where it wrongfully things the question is confused.
### What did Paul disagree with Peter about according to Paul's letter to the Galatians? Who was right?
The model's answer is mostly right. The text mentions that Peter was eating Gentile food, and then hypocritically saying that Gentiles must abstain from their food. I was looking for a mention of circumcision, which was a more fundamental concern of Paul's in Galations, as he saw circumcision as a legalistic curse for a Gentile to go through.
### Why did Peter feel comfortable eating pork and shellfish after receiving the Holy Spirit?
The model mentions Peter's vision of God cleaning impure food, which is what I was looking for. It does not seem able to draw the connection that pork and shellfish are some of the impure foods that Peter was allowed to eat. That would probably not be solved by a RAG system.
### When Jesus mentions that Moses lifted the serpent in the wilderness (John 3:14), what is He referencing?
The model correctly identifies the passage in the Book of Numbers that Jesus referenced.
### What is the genealogy of Enoch going back to Adam?
The genealogy is correct, as is the reference to Genesis 5.
### Why did God flood the world in the time of Noah?
The model correctly identifies the verse that explains why God wanted to flood the Earth.
### What are the similarities between Daniel's prophesies and Revelation's prophesies?
I was mainly looking for references to the Antichrist from Daniel & Revelation, which are given correctly.
