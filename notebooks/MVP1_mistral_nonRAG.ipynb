# Biblical RAG LLM MVP 1

This notebook tests mistral:instruct's ability to with no Bible support.

## Instantiate LLM

```python
from langchain_ollama import OllamaLLM

llm = OllamaLLM(model='mistral:instruct')
print("Model isntantiated.")
```

## Test the RAG Model

```python
questions = [
    "Where did Jesus talk about wine?",
    "Who was Abraham's son?",
    "Where did God send Jonah? And where did Jonah try to go instead?",
    "Who are Luke's books addressed to?",
    "Which psalm says that angels will protect you even from dashing your foot against a stone?",
    "Why did God regret making Saul the King of Israel?",
    "What did Paul disagree with Peter about according to Paul's letter to the Galatians? Who was right?",
    "Why did Peter feel comfortable eating pork and shellfish after receiving the Holy Spirit?",
    "When Jesus mentions that Moses lifted the serpent in the wilderness (John 3:14), what is He referencing?",
    "What is the genealogy of Enoch going back to Adam?",
    "Why did God flood the world in the time of Noah?",
    "What does Messiah mean in Hebrew?",
    "How do you spell 'Jesus' in Hebrew and Greek?",
    "What are the similarities between Daniel's prophesies and Revelation's prophesies?"
]
print("There are {} questions in the test suite.".format(len(questions)))

for question in questions:
    answer = llm.invoke(question)
    qa = {'question': question, 'answer': answer}
    print(qa)
```

## Conclusion

### Where did Jesus talk about wine?
The model get some correct answers and adds some passages where wine is not mentioned at all.
### Who was Abraham's son?
The model only mentions Isaac, which is acceptable as Ishmael was a bastard.
### Where did God send Jonah? And where did Jonah try to go instead?
The model correctly mentions Nineveh and Tarshish, and then provides possible locations of Tarshish (no one knows for sure where it was).
### Who are Luke's books addressed to?
The model wrongfully does not mention Theophilus.
### Which psalm says that angels will protect you even from dashing your foot against a stone?
The model correctly identifies Psalm 91, and that the reference is in verse 12.
### Why did God regret making Saul the King of Israel?
The model correctly identifies the specific event that lead God to reject Saul, and the passage that describes the event when Samuel declares it.
### What did Paul disagree with Peter about according to Paul's letter to the Galatians? Who was right?
The model correctly identifies that the reason for their disagreement was circumcision of Gentiles, and the passage that descibes this.
### Why did Peter feel comfortable eating pork and shellfish after receiving the Holy Spirit?
The model understands that pork and shellfish violate the Law of Moses and identifies the passage where Peter has the vision where God tells him all animals are clean.
### When Jesus mentions that Moses lifted the serpent in the wilderness (John 3:14), what is He referencing?
The model correctly identifies the story in the Book of Numbers and its meaning.
### What is the genealogy of Enoch going back to Adam?
The model correctly identifies the geneology.
### Why did God flood the world in the time of Noah?
The model correctly identifies the reason for the flood.
### What are the similarities between Daniel's prophesies and Revelation's prophesies?
The model provided explanations, but no Biblical references to their similarities.
